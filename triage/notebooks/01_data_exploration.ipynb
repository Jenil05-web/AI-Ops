## Data Loading & Cleaning Phase :

```python
import pandas as pd
import os
data_dir = '../data/raw'
files = ['dataset-tickets-multi-lang-4-20k.csv', 'dataset-tickets-multi-lang3-4k.csv', 'aa_dataset-tickets-multi-lang-5-2-50-version.csv']
df_temp = [pd.read_csv(os.path.join(data_dir, f)) for f in files]
df = pd.concat(df_temp, ignore_index=True)

## We loaded the csv file and we will process it according to our needs ( by removing certain unecessary things basically cleaning it to need and then saving it )
```
